```python
from openai import OpenAI
```

```python
openai = OpenAI()
```

```python
messages = [{"role": "user", "content": "What is 2 + 2?"}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages
)
print(response.choices[0].message.content)
```

```python
question = "Please propose a hard, challeinging questino to assess someones IQ. Respond only with the question, no other text."
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": question}]
)
print(response.choices[0].message.content)
```
