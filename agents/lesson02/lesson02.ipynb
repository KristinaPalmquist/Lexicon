```python
from openai import OpenAI
from Markdown import Markdown
messages = [{"role": "user", "content": "What is 2 + 2?"}]
```

```python
ollama = OpenAI(base_url="http://localhost:11434/v1", api_key='ollama')  # API key doesn't matter
model_name = "llama3.2"
response = ollama.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
print(competitors)
print(answers)
```

```python
for competitors, answer in zip(competitors, answers):
    print(f"# competitor: {competitors}\n\n{answer}")
```

```python
together = ""
for index, answer in enumerate(answers):
    together += f"# Response from competitor {index+1}"
    together += answer + "\n\n"
print(together)
```
