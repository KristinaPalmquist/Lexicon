```python
import tensorflow_datasets as tfds

# TFDS will auto-download the dataset the first time.
# We'll split the single "train" into train/val/test (80/10/10).
splits = {
    "train": "train[:80%]",
    "val":   "train[80%:90%]",
    "test":  "train[90%:]"
}

(ds_train, ds_val, ds_test), ds_info = tfds.load(
    "cats_vs_dogs",
    split=[splits["train"], splits["val"], splits["test"]],
    with_info=True,
    as_supervised=True,
)

num_classes = ds_info.features["label"].num_classes
class_names = ds_info.features["label"].names
print("Classes:", class_names, "| Num classes:", num_classes)
 
```
