## Import the data

```python
import pandas as pd
```

```python
female_dataset = pd.read_csv('./data/female_sized.csv')
female_dataset.shape
```

```python
male_dataset = pd.read_csv('./data/male_sized.csv')
male_dataset.shape
```

## Scikit-learn

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

male_X = male_dataset.drop(columns=['chestcircumference', 'biacromialbreadth', 't-shirt size'])
male_y = male_dataset['t-shirt size']
female_X = female_dataset.drop(columns=['chestcircumference', 'biacromialbreadth', 't-shirt size'])
female_y = female_dataset['t-shirt size']

k = 5
t = 0.2

male_X_train, male_X_test, male_y_train, male_y_test =train_test_split(male_X, male_y, test_size=t, random_state=42)

female_X_train, female_X_test, female_y_train, female_y_test =train_test_split(female_X, female_y, test_size=t, random_state=42)
```

```python
print('Original')
sk_clf = KNeighborsClassifier(n_neighbors=k)
sk_clf.fit(male_X_train, male_y_train)
predictions = sk_clf.predict(male_X_test)
accuracy = accuracy_score(male_y_test, predictions)
print(f'Male: k = {k}, t = {t}, Accuracy = {accuracy:.3f}')

sk_clf = KNeighborsClassifier(n_neighbors=5)
sk_clf.fit(female_X_train, female_y_train)
predictions = sk_clf.predict(female_X_test)
accuracy = accuracy_score(female_y_test, predictions)
print(f'Female: k = {k}, t = {t}, Accuracy = {accuracy:.3f}')
```

```python
print('Test Size')
k = 5
for t in [0.1, 0.2, 0.3, 0.5]:
    X_train, X_test, y_train, y_test = train_test_split(male_X, male_y, test_size=t, random_state=42)
    sk_clf = KNeighborsClassifier(n_neighbors=5)
    sk_clf.fit(X_train, y_train)
    predictions = sk_clf.predict(X_test)
    accuracy = accuracy_score(y_test, predictions)
    print(f'Male: k = {k}, t = {t}: Accuracy = {accuracy:.3f}')

for t in [0.1, 0.2, 0.3, 0.5]:
    X_train, X_test, y_train, y_test = train_test_split(female_X, female_y, test_size=t, random_state=42)
    sk_clf = KNeighborsClassifier(n_neighbors=5)
    sk_clf.fit(X_train, y_train)
    predictions = sk_clf.predict(X_test)
    accuracy = accuracy_score(y_test, predictions)
    print(f'Female: k = {k}, t = {t}: Accuracy = {accuracy:.3f}')
```

```python
print('Neighbors')
t = 0.2
for k in [1, 3, 5, 7, 11, 15, 21]:
    sk_clf = KNeighborsClassifier(n_neighbors=k)
    sk_clf.fit(male_X_train, male_y_train)
    predictions = sk_clf.predict(male_X_test)
    accuracy = accuracy_score(male_y_test, predictions)
    print(f'Male: k = {k}, t = {t}, Accuracy = {accuracy:.3f}')
```

```python
print('Neighbors')
for k in [1, 3, 5, 7, 11, 15, 21]:
    sk_clf = KNeighborsClassifier(n_neighbors=k)
    sk_clf.fit(female_X_train, female_y_train)
    predictions = sk_clf.predict(female_X_test)
    accuracy = accuracy_score(female_y_test, predictions)
    print(f'Female: k = {k}, t = {t}, Accuracy = {accuracy:.3f}')
```
